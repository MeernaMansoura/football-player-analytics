```python

# 1. أفضل 5 هدافين
top_scorers = df_merged.sort_values(by='Goals', ascending=False)[['Player', 'Club', 'Position', 'Goals']].head(5)

# 2. أفضل 5 صانعي ألعاب (Assists)
top_playmakers = df_merged.sort_values(by='Assists', ascending=False)[['Player', 'Club', 'Position', 'Assists']].head(5)

# 3. اللاعبون الأكثر مساهمة (أهداف + تمريرات حاسمة)
df_merged['Total_Contributions'] = df_merged['Goals'] + df_merged['Assists']
top_contributors = df_merged.sort_values(by='Total_Contributions', ascending=False)[['Player', 'Club', 'Position', 'Goals', 'Assists', 'Total_Contributions']].head(5)

print("--- أفضل 5 هدافين ---")
print(top_scorers.to_string(index=False))

print("\n--- أفضل 5 صانعي ألعاب ---")
print(top_playmakers.to_string(index=False))

print("\n--- أكثر اللاعبين مساهمة (أهداف + صناعة) ---")
print(top_contributors.to_string(index=False))
```

```python
df_salary_clean = df_merged.dropna(subset=['Salary'])
```

```python
top_earners = df_salary_clean.sort_values(by='Salary', ascending=False)[
    ['Player', 'Club', 'Salary', 'Goals', 'Total_Contributions']
].head(5)
```

```python
print("--- أعلى 5 لاعبين أجراً ومقارنة أهدافهم ---")
print(top_earners.to_string(index=False))
```

```python
correlation = df_salary_clean['Salary'].corr(df_salary_clean['Goals'])
print(f"\nمعامل الارتباط بين الراتب والأهداف: {correlation:.2f}")
```

```python
df_merged['Total_Contributions'] = df_merged['Goals'] + df_merged['Assists']

# 2. تنظيف البيانات: استبعاد القيم الفارغة في الرواتب واللاعبين بدون مساهمات
df_value = df_merged.dropna(subset=['Salary']).copy()
df_value = df_value[df_value['Total_Contributions'] > 0]

# 3. حساب تكلفة المساهمة الواحدة (Cost Per Contribution)
df_value['Cost_Per_Contribution'] = df_value['Salary'] / df_value['Total_Contributions']

# 4. الترتيب تصاعدياً (من الأقل تكلفة إلى الأعلى) لإظهار أفضل الصفقات
best_value_players = df_value.sort_values(by='Cost_Per_Contribution', ascending=True)[
    ['Player', 'Club', 'Salary', 'Total_Contributions', 'Cost_Per_Contribution']
].head(5)

print("--- أفضل 5 صفقات رابحة (أعلى تأثير بأقل تكلفة) ---")
print(best_value_players.to_string(index=False))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt


# 1. إعداد السمة والألوان
sns.set_theme(style="whitegrid", palette="deep")

# 2. ترتيب البيانات تنازلياً حسب العدد لسهولة القراءة
position_order = df_merged['Position'].value_counts().index

# 3. إنشاء الرسم البياني
plt.figure(figsize=(12, 7))
ax = sns.countplot(
    data=df_merged,
    x='Position',
    order=position_order,
    color= 'purple',
    edgecolor="black"  # إضافة حدود للأعمدة
)

# 4. إضافة العناوين وتسميات المحاور
plt.title('Distribution of Players by position', fontsize=16, fontweight='bold')
plt.xlabel('Position', fontsize=12)
plt.ylabel('Number of players', fontsize=12)

# 5. تدوير تسميات المحور السيني إذا كانت طويلة
plt.xticks(rotation=45)

# 6. إضافة أرقام القيم فوق كل عمود
for p in ax.patches:
    ax.annotate(format(p.get_height(), '.0f'), 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha = 'center', va = 'center', 
                xytext = (0, 9), 
                textcoords = 'offset points')

plt.tight_layout()
plt.show()
```

```python
# رسم توزيع الرواتب مع كشف القيم الشاذة


sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 6))
sns.boxplot(
    y=df_salary_clean['Salary'],
    color='purple',        # اللون النهدي
    fliersize=5            # حجم نقاط القيم الشاذة
)

plt.title('Salary Distribution and Outliers', fontsize=14, fontweight='bold', pad=15)
plt.ylabel('Salary', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_salary_clean, 
    x='Salary', 
    y='Goals',
    color='purple',        # اللون النهدي
    s=70,                  # حجم النقاط لتكون أضح وأجمل
    alpha=0.7              # شفافية خفيفة لتمييز التداخلات
)

plt.title('Relationship Between Salary and Goals', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Salary', fontsize=12)
plt.ylabel('Goals', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# ترتيب المراكز تنازلياً حسب متوسط الراتب ليكون الرسم منظماً
salary_order = df_merged.groupby('Position')['Salary'].mean().sort_values(ascending=False).index

plt.figure(figsize=(12, 6))
sns.barplot(
    data=df_merged, 
    x='Position', 
    y='Salary', 
    order=salary_order,
    color='purple',        # اللون النهدي
    edgecolor="black"      # حدود سوداء خفيفة لإبراز الأعمدة
)

plt.title('Average Salary by Position', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Position', fontsize=12)
plt.ylabel('Average Salary', fontsize=12)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
